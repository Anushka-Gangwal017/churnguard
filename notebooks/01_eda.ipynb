### Churn Distribution

Approximately 73.46% of customers did not churn, while 26.54% of customers churned.
This shows that the dataset is imbalanced, with fewer customers in the churn class.
This class distribution should be considered during model development and evaluation.

```python
sns.countplot(data=df, x="Churn")

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

```python
sns.boxplot(data=df, x="Churn", y="tenure")

plt.title("Tenure vs Churn")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.show()
```

```python
sns.boxplot(data=df, x="Churn", y="MonthlyCharges")

plt.title("Monthly Charges vs Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.show()
```

```python
sns.countplot(data=df, x="Contract", hue="Churn")

plt.title("Contract Type vs Churn")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=15)
plt.show()
```

```python
sns.countplot(data=df, x="InternetService", hue="Churn")

plt.title("Internet Service vs Churn")
plt.xlabel("Internet Service")
plt.ylabel("Number of Customers")
plt.show()
```

## EDA Summary

- The dataset contains 7043 customer records and 21 columns.
- There are no duplicate rows in the dataset.
- The `Churn` variable contains 73.46% non-churned customers and 26.54% churned customers.
- Customers who churned generally have lower tenure than customers who stayed.
- Customers who churned generally have higher monthly charges.
- Month-to-month contract customers show considerably higher churn than customers with one-year or two-year contracts.
- `TotalCharges` is stored as a text column and contains 11 blank values, which will be handled during preprocessing.
